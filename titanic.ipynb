## Project: Titanic Survival Analysis
This project explores the Titanic dataset to understand what factors influenced passenger survival.

## 1. Data Overview

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")
df.head()
```

```python
df.info()
df.describe()
df.isnull().sum()
```

```python
df.info()
df.describe()
df.isnull().sum()
```

## 2. Data Cleaning

```python
# Fill missing Age and Embarked
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Drop 'Cabin' if it exists
if 'Cabin' in df.columns:
    df = df.drop(columns=['Cabin'])
```

## 3. Exploratory Data Analysis (EDA)

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Survival count
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.show()

# Survival by Sex
sns.countplot(x='Survived', hue='Sex', data=df)
plt.title("Survival by Gender")
plt.show()

# Age Distribution
sns.histplot(df['Age'], kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
# Convert 'Sex' to numeric for correlation
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# Heatmap
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Feature Correlation")
plt.show()
```
